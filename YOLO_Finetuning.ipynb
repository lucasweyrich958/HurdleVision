```python
# Step 1: Install the necessary libraries
!pip install -U ultralytics wandb

# Step 3: Run your training as usual
from ultralytics import YOLO

modeln = YOLO('yolo11n')
models1 = YOLO('yolo11s')
models2 = YOLO('yolo11s')
models3 = YOLO('yolo11s')
modelm = YOLO('yolo11m')

import wandb

# Initialize your Weights & Biases environment
wandb.login(key="")
```

### Nano Model

```python
!yolo settings wandb=True
# Train the model with augmentations
results = modeln.train(
   data='/content/Annotated Frames/hurdles.yaml',
   epochs=100,
   imgsz=640,
   batch=10,
   pretrained=True,
   plots=True,
   optimizer='AdamW',
   lr0=0.0001,
   # --- Geometric Augmentations ---
   degrees=7.0,        # Small rotation to simulate slight camera tilt.
   translate=0.2,      # Good for objects appearing in different parts of the frame.
   scale=0.5,          # Very beneficial, as athletes/hurdles are at different distances.
   shear=3.0,          # Small shear is okay for perspective shifts; avoid large, unnatural distortions.
   perspective=0.0,    # Keep this off unless you have significant camera perspective changes.
   fliplr=0.5,         # HORIZONTAL FLIP: Highly recommended. A race can be filmed from the left or right.
   flipud=0.0,         # VERTICAL FLIP: MUST be disabled. Athletes and hurdles are never upside down.

   # --- Color Augmentations (Very Important) ---
   hsv_h=0.015,        # Adjusts hue. Good for different track colors and lighting.
   hsv_s=0.7,          # Adjusts saturation. Simulates different broadcast color grading.
   hsv_v=0.4,          # Adjusts brightness. Crucial for day vs. night races.

   # --- Structural Augmentations ---
   mosaic=1.0,         # Helps the model handle occlusion (athletes blocking each other).
   mixup=0.1,          # Good regularizer.
   copy_paste=0.1,          # IMPORTANT: Ensures images are processed one by one
   name = 'yolo11n_run',
   project = 'HurdleVision'
)
```
